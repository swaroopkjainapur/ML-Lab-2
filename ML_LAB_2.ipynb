```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. LOAD THE DATASET
#    Using the Iris dataset for classification example.
# ------------------------------------------------------------
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['species'] = iris.target

print("Iris dataset loaded successfully!")
print("\nFirst 5 rows of the dataset:")
display(df.head())

print("\nShape of the dataset:")
print(df.shape)

print("\nColumn names:")
print(df.columns)

print("\nData types:")
print(df.dtypes)

print("\nStatistical Summary:")
display(df.describe())

print("\nMissing values count per column:")
print(df.isnull().sum())

# ------------------------------------------------------------
# 2. TARGET VARIABLE DISTRIBUTION
# ------------------------------------------------------------
print("\nCounts of each Iris species (target variable):")
print(df["species"].value_counts())

plt.figure(figsize=(7, 5))
sns.countplot(x="species", data=df)
plt.title("Distribution of Iris Species")
plt.xlabel("Species (0: setosa, 1: versicolor, 2: virginica)")
plt.ylabel("Count")
plt.show()

# ------------------------------------------------------------
# 3. FEATURE DISTRIBUTION: SEPAL LENGTH
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
sns.histplot(df["sepal length (cm)"], bins=10, kde=True)
plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Frequency")
plt.show()

# ------------------------------------------------------------
# 4. SEPAL LENGTH VS SPECIES
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
sns.boxplot(x="species", y="sepal length (cm)", data=df)
plt.title("Sepal Length vs. Iris Species")
plt.xlabel("Species (0: setosa, 1: versicolor, 2: virginica)")
plt.ylabel("Sepal Length (cm)")
plt.show()

# ------------------------------------------------------------
# 5. PETAL LENGTH VS SPECIES
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
sns.boxplot(x="species", y="petal length (cm)", data=df)
plt.title("Petal Length vs. Iris Species")
plt.xlabel("Species (0: setosa, 1: versicolor, 2: virginica)")
plt.ylabel("Petal Length (cm)")
plt.show()

# ------------------------------------------------------------
# 6. PETAL WIDTH VS SPECIES
# ------------------------------------------------------------
plt.figure(figsize=(7, 5))
sns.boxplot(x="species", y="petal width (cm)", data=df)
plt.title("Petal Width vs. Iris Species")
plt.xlabel("Species (0: setosa, 1: versicolor, 2: virginica)")
plt.ylabel("Petal Width (cm)")
plt.show()

# ------------------------------------------------------------
# 7. CORRELATION HEATMAP OF FEATURES
# ------------------------------------------------------------
plt.figure(figsize=(10, 8))
sns.heatmap(
    df.drop('species', axis=1).corr(), # Exclude target for feature correlation
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap of Iris Features")
plt.show()

# ------------------------------------------------------------
# 8. SEPARATE FEATURES AND TARGET VARIABLE
# ------------------------------------------------------------
X = df.drop("species", axis=1) # Features are all columns except 'species'
y = df["species"]          # Target is the 'species' column

print("\nFeatures (X) - first 5 rows:")
display(X.head())

print("\nTarget (y) - first 5 rows:")
display(y.head())

# ------------------------------------------------------------
# 9. SPLIT DATA INTO TRAINING AND TESTING SETS
# ------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2, # 20% of data for testing
    random_state=42, # For reproducibility
    stratify=y # Ensure equal proportion of classes in train/test splits
)

print("\nTraining data size:", X_train.shape)
print("Testing data size:", X_test.shape)

# ------------------------------------------------------------
# 10. INITIALIZE THE CLASSIFIER MODEL
# ------------------------------------------------------------
model = DecisionTreeClassifier(random_state=42)

# ------------------------------------------------------------
# 11. TRAIN THE MODEL
# ------------------------------------------------------------
model.fit(X_train, y_train)

print("\nModel training completed successfully!")

# ------------------------------------------------------------
# 12. MAKE PREDICTIONS ON THE TEST SET
# ------------------------------------------------------------
y_pred = model.predict(X_test)

print("\nModel predictions on the test set:")
print(y_pred)

# ============================================================
# MODEL EVALUATION
# ============================================================

print("\n====================================")
print("       MODEL EVALUATION")
print("====================================")

# ------------------------------------------------------------
# 13. CALCULATE EVALUATION METRICS
# ------------------------------------------------------------
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted') # Use 'weighted' for multiclass
recall = recall_score(y_test, y_pred, average='weighted')     # Use 'weighted' for multiclass
f1 = f1_score(y_test, y_pred, average='weighted')             # Use 'weighted' for multiclass

print("Accuracy  :", round(accuracy, 2))
print("Precision :", round(precision, 2))
print("Recall    :", round(recall, 2))
print("F1 Score  :", round(f1, 2))

# ------------------------------------------------------------
# 14. CONFUSION MATRIX VISUALIZATION
# ------------------------------------------------------------
cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(cm)

plt.figure(figsize=(7, 6))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=iris.target_names,
    yticklabels=iris.target_names
)
plt.title("Confusion Matrix")
plt.xlabel("Predicted Species")
plt.ylabel("Actual Species")
plt.show()

# ------------------------------------------------------------
# 15. PERFORMANCE METRICS BAR CHART
# ------------------------------------------------------------
metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1 Score"
]
scores = [
    accuracy,
    precision,
    recall,
    f1
]

plt.figure(figsize=(8, 6))
bars = plt.bar(metrics, scores, color=['skyblue', 'lightcoral', 'lightgreen', 'gold'])
plt.ylim(0, 1.1) # Set y-axis limit slightly above 1 for text labels
plt.ylabel("Score")
plt.title("Model Performance Metrics")

for bar, score in zip(bars, scores):
    plt.text(
        bar.get_x() + bar.get_width() / 2, # Center text above bar
        score + 0.05, # Position text slightly above bar
        f"{score:.2f}",
        ha="center",
        va="bottom"
    )
plt.show()

# ------------------------------------------------------------
# 16. DECISION TREE VISUALIZATION
# ------------------------------------------------------------
plt.figure(figsize=(20, 10))
plot_tree(
    model,
    filled=True,
    feature_names=X.columns, # Use actual feature names
    class_names=iris.target_names, # Use actual species names
    max_depth=3, # Limit depth for better visualization
    fontsize=10
)
plt.title("Decision Tree Visualization for Iris Classification")
plt.show()

# ------------------------------------------------------------
# 17. FINAL SUMMARY OF RESULTS
# ------------------------------------------------------------
print("\n====================================")
print("           FINAL RESULTS")
print("====================================")
print("Dataset size       :", df.shape)
print("Training samples   :", len(X_train))
print("Testing samples    :", len(X_test))
print("Accuracy           :", round(accuracy, 2))
print("Precision (weighted):", round(precision, 2))
print("Recall (weighted)   :", round(recall, 2))
print("F1 Score (weighted) :", round(f1, 2))
```
